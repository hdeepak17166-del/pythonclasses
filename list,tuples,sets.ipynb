```python
#list
students=[35,45,50,60,70] #can store same datatype data
print(type(students))
students1=[34,'mango',37.5,True] #can store multipal datatype data
print(type(students1))
```

```python
#indexing in list
fruits=['banana','apple','mango']
print(fruits[0])
print(fruits[-1])
```

```python
#slicing in list
#[start:stop:step]
fruits=['banana','apple','mango']
print(fruits[1:3])
print(fruits[0:2])
print(fruits[:])
print(fruits[::2])
print(fruits[::-1]) #can also use for reversing the text
```

```python
#list are mutable{change}in nature
list2=[10,20,30,40]
list2[0]=100
print(list2) #o replaces the 10 to 100  
```

```python
#repeating list
list3=[1]*5
print(list3)
```

```python
#list methods - .append, .insert
list2=[10,20,30,40]
list2.append(50) # append is adding the value to the end of the list

list2=[10,20,30,40]
list2.append([50,60,70])
print(list2)

list2.insert(1,35) # insert is to add a value using index position
print(list2)
list2.remove(35) #exact value to removed from the list
print(list2)
list2.pop() # by default remove the last value
print(list2)
list2.pop(1) #removing the element using index position
print(list2)
```

```python
a=[1,2,3]
b=[4,5,6]
a.extend(b) #to extend the eisting two variable
print(a)
```

```python
mark=[78,102,96]
mark.sort() #sort is to arrange the list according to the assending order
print(mark)
mark.reverse() #reverse is to reverse  the order of the elements in the list
print(mark)
mark.clear() #to remove everthing
print(mark)
```

```python
nums=[10,20,78,89]
print(nums.index(78)) #ive the positions of the item
nums2=[1,2,3,3,2,2]
print(nums2.count(2)) #counting occurrences

a=[1,2,3]
b=a.copy() #creating a shallow copy
print(b)
print(len(nums2)) #lenght of the list
```

```python
#deepcopy full copy, for deepcopy we import copy module because it's not a inbult method

import copy
a=[[1,2],[3,4]]
b=copy.deepcopy(a)
b[0][0]=100
print(a)
print(b)
```
