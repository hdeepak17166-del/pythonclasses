```python
#list
students=[35,45,50,60,70] #can store same datatype data
print(type(students))
students1=[34,'mango',37.5,True] #can store multipal datatype data
print(type(students1))
```

```python
#indexing in list
fruits=['banana','apple','mango']
print(fruits[0])
print(fruits[-1])
```

```python
#slicing in list
#[start:stop:step]
fruits=['banana','apple','mango']
print(fruits[1:3])
print(fruits[0:2])
print(fruits[:])
print(fruits[::2])
print(fruits[::-1]) #can also use for reversing the text
```

```python
#list are mutable{change}in nature
list2=[10,20,30,40]
list2[0]=100
print(list2) #o replaces the 10 to 100  
```

```python
#list methods - .append, .insert
list2=[10,20,30,40]
list2.append(50) # append is adding the value to the end of the list
print(list2)
list2.insert(1,35) # insert is to add a value using index position
print(list2)
list2.remove(35) #exact value to removed from the list
print(list2)
list2.pop() # by default remove the last value
print(list2)
list2.pop(1) #removing the element using index position
print(list2)
```
