Structural pattern matching(match-case)
Match-case is python's switch statement 

use it when:
you compare one variable against many fixed values

Basic syntax:
case option1:
    #code 
case option2:
    #code
case option3:
    #code
case _:
#default

Here _ means "anything else"

```python
#menu seletion

choice=2                       #can also wright - int(input("enter your choice"))

match choice:                
    case 1:
        print("Ad item")
    case 2:
        print("Delete Item")
    case 3:
        print("Update the item")
    case _:
        print("invalid choice")
```

```python
# using strings

day="sunday"

match day:
    case "saturday"|"sunday":
        print("weekend")
    case _:
        print("weekday")
```

```python
#simple calculater option

operation="+"

match operation:                              #match -> check this value
    case "+":                                 #case -> if it matches this
        print("addition selected")            #_ -> anything else
    case "-":
        print("subtraction selected")
    case "/":
        print("division selected")
    case "*":
        print("multiplication selected")
    case _:
        print("invalied operation")
```

Match is for the condition like greater than or less than
