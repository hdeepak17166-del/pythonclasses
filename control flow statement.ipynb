Conditionals(if,elif,else)
Conditional lets the programme make decision
"if somethin is true-> do this . otherwie do-> that"

Basic syntax-
if condition:
run if condition is true 

elif another_condition:
run if above is false and this is true 

else:
rune if all conditions are false

```python
#simple if statement
age=20
if age>=18:
    print("you can vote")
```

```python
# if-else statatement
rain=True

if rain:
    print("take umbrella")
else:
    print("dont take umbrella")
```

```python
# one more example for if-else

marks=60
if marks>=40:
    print("pass")
else:
    print("fail")
```

```python
# elif statement

weather="sunny"

if weather=="rainy":
    print("take umbrella")
elif weather=="sunny":
    print("take sunglasses")
else:
    print("it is a pleasent day")
```

```python
# more examples for if- elif- else

score=78

if score>=90:
    print("Grade A")
elif score>=75:
    print("Grade B")
elif score>=60:
    print("Grade C")
else:
    print("Grade D")
```

Structural pattern matching(match-case)
Match-case is python's switch statement 

use it when:
you compare one variable against many fixed values

Basic syntax:
case option1:
    #code 
case option2:
    #code
case option3:
    #code
case _:
#default

Here _ means "anything else"

```python
#menu seletion

choice=2                       #can also wright - int(input("enter your choice"))

match choice:                
    case 1:
        print("Ad item")
    case 2:
        print("Delete Item")
    case 3:
        print("Update the item")
    case _:
        print("invalid choice")
```

```python
# using strings

day="sunday"

match day:
    case "saturday"|"sunday":
        print("weekend")
    case _:
        print("weekday")
```

```python
#siple calculater option

operation="+"

match operation:                              #match -> check this value
    case "+":                                 #case -> if it matches this
        print("addition selected")            #_ -> anything else
    case "-":
        print("subtraction selected")
    case "/":
        print("division selected")
    case "*":
        print("multiplication selected")
    case _:
        print("invalied operation")
```

Match is for the condition like greater than or less than
Match cannot be used for a range of values either
