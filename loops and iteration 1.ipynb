FOR LOOP

A for loop takes one item at a time form a collection and work on it.

syntax-
for item in collection:
    print(item)

collection-list/string/range
item-temporary variable

```python
# basic example

for i in range(5):
    print(i)

# range(5) -> numbers from 0 to 4
#i takes one value at a time
#print(i) runs repeatedly until the for loop condition is true
```

```python
#print number 1 t 6

for i in range(1,6):
    print(i)
```

```python
# for loop with list item
fruits=["apple","bannana","mango"]
for i in fruits:
    print(i)
```

```python
#loop through a string

name="Python"

for letter in name:
    print(letter)

#strings are collections of characters
```

```python
#number list

numbers=[10,20,30]

for n in numbers:
    print(n*2)
```

```python
#for loop with a dictionary

students={"name":"deepak","age":23}

for value in students.values():
    print(value)
```

Checklist to decide which loop to use -
1. Use for if:
- you have a list/string/dictionary
- you are looping through rows or values
- The data controls how many times to run

2. Use while if :
- You are waiting for condition to change 
- User decides when to stop
- you are validating an input 
- You don't know many repetition are needed

for-loop=data-driven repetition
While-loop=condition-driven repetition

5. repeat the menu until the user chooses exit - > While 
6. Read marks form the list print pass or fail for each student -> for 
7. remove items form the list until it becomes empty-> while

control flow keywords

Break- stop loop completely
brak immediately exits the loop.
nce break runs , the loop stops forever 

Continue- Skip the current iteration
continue skips the current loop step and ,moves to the next one.
the loop does not stop

Pass- pass does nothing . It is used when Python expect's code but you're not ready to write it.
yet. used like a placeholder
the loop does not stop here also.

Break- > atop the loop
Continue-> skip the step
Pass -> do nothing

```python
#break, continue and pass

#Break 
for i in range(5):
    if i ==3:
        break
    print(i)
```

```python
#Continue
for i in range(5):
    if i ==3:
        continue
    print(i)
```

```python
#Pass
for i in range(5):
    if i ==3:
        pass
    else:
        print(i)
```

# list comprehensaions 

[x for x in list if condition]

List comprehension is a shorter way to create a new list using logic 

```python
#normal way

numbers=[1,2,3]
squares=[]

for n in numbers:
    squares.append(n*n)

print(numbers)
print(squares)  
```

```python
#list comprehension way
numbers=[1,2,3]
squares=[n*n for n in numbers]

print(numbers)
print(squares)
```

```python
#with condition
evens=[n for n in numbers if n%2==0]
print(evens)
```

```python
#dictionary comprehension
numbers=[1,2,3]
result={}

for n in numbers:
    result[n]=n*n
print(result)
```

```python
#comprehension way

result={n:n*n for n in numbers}
print(result)
```

```python
# with condition

even_squares={n:n*n for n in numbers if n%2==0}
print(even_squares)
```

NESTED LOOPS
A LOOP WITHIN ANOTHER LOOP

use when data contains data inside it or when rows contains column

```python
#example

data=[
    [10,20],
    [30,40]
]

for row in data:
    for value in row:
        print(value)

#outer loop is for rows here 
#inner loop is for values inside each row 
```

```python
# totla of each row 

for row in data:
    total=0
    for value in row:
        total+=value 
    print(total)
```
