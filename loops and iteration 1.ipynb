Loops-
A loop helps us to repeate the same task again and again without writing the same code many types

example-
writing attendance for 30 students
printing numbers from 1 to 100
checking marks for each subject

types of loops-
for loop
while loop

FOR LOOP

A for loop takes one item at a time form a collection and work on it.

syntax-
for item in collection:
    print(item)

collection-list/string/range
item-temporary variable

```python
# basic example

for i in range(5):
    print(i)

# range(5) -> numbers from 0 to 4
#i takes one value at a time
#print(i) runs repeatedly until the for loop condition is true
```

```python
#print number 1 t 6

for i in range(1,6):
    print(i)
```

```python
# for loop with list item
fruits=["apple","bannana","mango"]
for i in fruits:
    print(i)
```

```python
#loop through a string

name="Python"

for letter in name:
    print(letter)

#strings are collections of characters
```

```python
#number list

numbers=[10,20,30]

for n in numbers:
    print(n*2)
```

```python
#for loop with a dictionary

students={"name":"deepak","age":23}

for value in students.values():
    print(value)
```

```python
students={"name":"deepak","age":23}

for key in students:
    print(key)
```

```python
#looping through key-value pairs
for key,value in students.items():
    print(key,":",value)
```

```python
#print even numbers

for i in range(0,11,2):       #0,11,2 - start, stop, step
    print(i)
```

```python
sales = [1000,2000,3000,4000]
total = 0


for amount in sales:
    total+= amount    #+= will only work when the total is defined 
print(total)
```

```python
items=["apple","banana","mango","grapes"]
count=0

for item in items:
    count=count+ 1
print(count)
```

```python
scores = [85, 92, 78, 90, 88]

total_score = 0
count = 0

for s in scores:
    total_score += s  
    count += 1        
average_score = total_score / count
print("Total Score:", total_score)
print("Average Score:", average_score)
```

```python
# count only items greater than 50

items=[50,60,70,30,40]
count=0

for item in items:
    if item>50:
        count+=1
print(count)
```

```python
#count only even numbers

numbers=[1,2,3,4,5,6]
count=0

for num in numbers:
    if num % 2== 0:
        count+=1
print(count)
```

```python
#count non-empty value

data=["apple","","mango","","banana"]
count=0
for item in data:
    if item:         
        count += 1
print("The number of non-empty items is:", count)
```

```python
#print high or low
#print high for sales>100 low for sales<100

sales=[80,150,60,200]
for s in sales:
    if s > 100:
        print(s, "is high")
    else:
        print(s, "is low")
```

```python
#replace values
#replace 0 with missing

data=[10,0,20,0,30]
cleaned_data=[]

for d in data:
    if d ==0:
        cleaned_data.append("missing")
    else:
        cleaned_data.append(d)

print(cleaned_data)
    
```
